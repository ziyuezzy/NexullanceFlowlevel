```python
import os
import sys
sys.path.append("/users/ziyzhang/topology-research")
from topologies.DDF import DDFtopo
import global_helpers as gl
import numpy as np
import pickle
```

```python
routing_table = pickle.load(open("./MD_NEXU_IT_all_shifts_(36,5)DDFtopo_paths.pickle", 'rb'))
```

```python
config = gl.ddf_configs[0]
V = config[0]
D = config[1]
EPR = (D+1)//2
_network = DDFtopo(V, D)
arcs = _network.generate_graph_arcs()

_network.pre_calculate_ECMP_ASP()

Cap_core = 10 #GBps
Cap_access = 10 #GBps
```

```python
for _shift in range(1, V*EPR):
    M_EP = gl.generate_shift_traffic_pattern(V, EPR, _shift)

    # try to scale the traffic scaling factor to 10x saturation under ECMP_ASP
    core_link_flows, access_link_flows = _network.distribute_M_EPs_on_weighted_paths(_network.ECMP_ASP, EPR, M_EP)
    max_core_link_load = np.max(core_link_flows)/Cap_core
    max_access_link_load = np.max(access_link_flows)/Cap_access
    traffic_scaling = 10.0/max(max_access_link_load, max_core_link_load)
    M_EP = traffic_scaling * M_EP

    core_link_flows, access_link_flows = _network.distribute_M_EPs_on_weighted_paths(routing_table, EPR, M_EP)
    max_access_link_load = np.max(access_link_flows)/Cap_access
    max_core_link_load = np.max(core_link_flows)/Cap_core

    print(f"shift_{_shift} traffic demand, Phi = {gl.network_total_throughput(M_EP, max_core_link_load, max_access_link_load)/(V*EPR)}")
```
